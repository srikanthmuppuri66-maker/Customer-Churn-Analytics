```python
import pandas as pd
import numpy as np

# 1. Load raw dataset
df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")

# 2. Convert spaces in TotalCharges to NaN and convert to float
df['TotalCharges'] = df['TotalCharges'].replace(" ", np.nan)
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'])

# 3. Fill missing values in TotalCharges with the median
df['TotalCharges'].fillna(df['TotalCharges'].median(), inplace=True)

# 4. Create a binary Churn_Numeric column (Yes = 1, No = 0)
df['Churn_Numeric'] = df['Churn'].apply(lambda x: 1 if x == 'Yes' else 0)

# 5. Drop duplicate rows
df.drop_duplicates(inplace=True)

# 6. Save the clean dataset
df.to_csv("cleaned_customer_churn.csv", index=False)
print("SUCCESS: cleaned_customer_churn.csv has been created!")
```
